```python
from sqlalchemy import create_engine, text
username= 'root'
password= 'Ridhima@123'
database= 'sql_project'
host= 'localhost'
engine = create_engine(f'mysql+pymysql://{username}:{password}@{host}/{database}')
# execute q query
connection = engine.connect()
result = connection.execute(text("select * from airport"))
for row in result:
    print(row)
connection.close()
```
