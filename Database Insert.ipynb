```python
from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.orm import declarative_base, sessionmaker
```

```python
DATABASE_URL='mysql+pymysql://root:MyNewPasswordadmin!@localhost/pw_test'
engine = create_engine(DATABASE_URL, echo=True)
```

```python
Base = declarative_base()
```

```python
#define emplyoee table structure
class Employee(Base):
    __tablename__='employees'
    id= Column(Integer,primary_key=True, autoincrement=True)
    name=Column(String(50),nullable=False)
    department= Column(String(50), nullable=False)
    sales_amount=Column(Float, nullable=False)
```

```python
Base.metadata.create_all(engine)
```

```python
Session = sessionmaker(bind=engine)
```

```python
with Session() as session:
    #Create employee objects
    employee_1 = Employee(name="Alice Brown",department="Sales",sales_amount=1200.00)
    employee_2 = Employee(name="Bob Clark",department="It",sales_amount=1700.60)
    employee_3 = Employee(name="Charlie",department="HR",sales_amount=1800.00)
    employee_4 = Employee(name="David",department="HR",sales_amount=500.00)
    employee_5 = Employee(name="Eve",department="Sales",sales_amount=2000.60)

    
    session.add_all([employee_1,employee_2,employee_3,employee_4,employee_5])
    session.commit()
    print("Records inserted successfully")
```

```python
with Session() as session:
    employees = session.query(Employee).all()
    for employee in employees:
        print(f"ID: {Employee.id}, Name: {employee.name}, Departments: {employee.department}, Sales Amount: {employee.sales_amount}")
```
